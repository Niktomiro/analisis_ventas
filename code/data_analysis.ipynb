```python
import numpy as np
import datetime as dt

def cargar_datos(ruta_archivo):
# Carga los datos del archivo CSV utilizando NumPy
    datos = np.genfromtxt(ruta_archivo, delimiter=',', skip_header=1, dtype=None)
    return datos

def head(datos):
    print("ID, date, Customer_ID, Gender, Age, PRoduct, Quantity, Price, Total amount")
    for i in range(10):
        print(f"{datos[i]}")

if __name__ == "__main__":
    ruta_archivo = '../data/retail_sales_dataset.csv'
    datos = cargar_datos(ruta_archivo)
    #print(datos)

#Columns: Transaction ID, Date, Customer_ID, gender, Age, Product category, Quantity, Price per unit, Total amount
head(datos)
print(datos.shape)
#Total de ventas por categoría de producto:
categories = []

for venta in datos:
    if str(venta[5]) not in categories:
        categories.append(str(venta[5]))

print(categories)
ventas = []

for cat in categories:
    for venta in datos:
        if str(venta[5]) == cat:
            ventas.append(venta[8])
    print(f"La venta total de la categoría: {cat} es: ${int(np.sum(ventas))}")
    ventas = []

# Ejercicio de promedio de fechas por categoría:

ventas_por_dia = []

for venta in datos:
    ventas_por_dia.append([str(venta[1]), str(venta[5]), int(venta[8])])

ventas_por_dia = np.array(ventas_por_dia)

for cat in categories:
    print(np.sum(np.where(str(ventas_por_dia[:, 1] == cat), ventas_por_dia[2], 0)))
#  print(f"El promedio de ventas diarias por {cat} es: {np.mean}")



```
